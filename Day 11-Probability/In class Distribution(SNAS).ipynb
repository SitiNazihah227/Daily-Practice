```python
# what is the average bill amount

sum_bill = df['total_bill'].sum()
count_bill = df['total_bill'].__len__()

print('The average bill amount is $', (sum_bill/count_bill))

print(sum_bill)
print(count_bill)
```

```python
sum_tip = df['tip'].sum()
count_tip = df['tip'].__len__()

print('The average tip amount is $', (sum_tip/count_tip))

print(sum_tip)
print(count_tip)
```
