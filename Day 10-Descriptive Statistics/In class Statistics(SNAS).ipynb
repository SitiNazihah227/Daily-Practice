```python
nums = [20,80,130,50,89,90,50,31,67,3]

def to_sort(nums):

    nums_length = len(nums)     #SNAS: first, to find the length of the arguments in the list

    for i in range(nums_length):
        smallest_num = i        #SNAS: try to create a placement to hold the smallest number in the list
        
        for j in range(i + 1, nums_length):     #SNAS: the next number in the range
            if nums[j] < nums[smallest_num]:    #SNAS: if the next number is smaller than the smallest number, assigned it that place. compares values using indexes
                smallest_num = j
        #SNAS: 

        nums[i], nums[smallest_num] = nums[smallest_num], nums[i]   #SNAS: swap the numbers

    return nums

sorted_nums = to_sort(nums)
print(f'The sorted list is: ', sorted_nums)
```

```python
nums = [20,80,130,50,89,90,50,31,67,3]

i = 0   #SNAS: first, indicate first index 0 is the smallest number
while i < len(nums):    #SNAS:  while the loop is still within the length of the list
    min_index = i   #SNAS: indicate the smallest number as initial number

    j = i + 1   #SNAS: increase the index position
    while j < len(nums):    #SNAS: while the increased loop is still within the list length
        if nums[j] < nums[min_index]:   #SNAS: compare the value at increased index with the value at current index
            min_index = j   #SNAS: if value at increased index is smaller than the value at the current index, update the numbervalue at that current index with increased index
        j += 1 #SNAS: continue loop at the next index

    #SNAS: swap the smallest value found with nums[i]
    nums[i], nums[min_index] = nums[min_index], nums[i]

    i += 1  #SNAS: outer loop, move to the next index

print(f'The sorted list is: ', nums)
```
