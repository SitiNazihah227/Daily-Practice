```python
newdf = df.copy()

median_price = newdf['Price'].median()  #SNAS: another method to fill in NaN values/missing values is to fill it with median.
print(median_price)

newdf['Price'].fillna(median_price, inplace=True) #SNAS: whichever has NaN values will be replaced with median value
newdf.head()
```

```python
threshold = 3   #SNAS: we will remove outliers. so in this example, whichever less than or more than 3, will be removed from the plot
 
z_scores = stats.zscore(df['Price'])
 
newdf_without_outliers = df[  (-threshold< z_scores) & (z_scores < threshold)  ]
```
