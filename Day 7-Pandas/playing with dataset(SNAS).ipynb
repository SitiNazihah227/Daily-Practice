```python
# calculated column

sepalarea = iris_without_species['sepal_length'] #SNAS: dataframe only takes one argument. 

sepalarea = iris_without_species[['sepal_length', 'sepal_width']]   #SNAS: to take more argument, must put it in a list, then it will take it as a dataframe

sepalarea.head()
```
