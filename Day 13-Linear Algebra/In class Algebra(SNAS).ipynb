```python
A = np.array(
[[2, 3, 4],
[6, 7, 8],
[3, 2, 3]])

#SNAS: A shape = 3 x 3

B = np.array([
[5, 5, 5],
[2, 2, 2],])

#SNAS: B shape = 2 x 3

B@A #SNAS: the result shape will be 2 x 3
```

```python
A = np.array(
[[5, 5, 5],
[2, 2, 2],
[5, 5, 5]])

B = np.array(
[[3, 3, 3],
[3, 3, 3],
[3, 3, 3]])

print('Matrix multiplication A@B: \n', A@B, '\n')
print('Array multiplication A*B: \n', A*B)
print('_________________________')
print('Matrix multiplication B@A: \n', B@A, '\n')
print('Array multiplication B*A: \n', B*A)
```
