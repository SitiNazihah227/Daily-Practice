```python
df.describe().round(2)  #SNAS: .round(2) rounding the number into 2 decimels
#SNAS: .describe giving info like below of the dataset
```

```python
df[['total_bill','tip','size']].head()  #SNAS: display the intended columns instead
```

```python
df[['total_bill','tip','size']].cov().round()  #SNAS: display the covariance of intended columns in rounded number

#SNAS: below is another way to display it
columns = ['total_bill','tip','size']
covariance = df[columns].cov()
covariance.round(2)
```

**Covariance Matrix**

- total bill vs. total bill = 79.25
- tip vs tip = 1.91
- size vs size = 0.90

**Correlation**

```python
correlation = df[columns].corr() #SNAS: corr() is the function to return correlation
correlation.round(2)

#SNAS: if there is any value that is negative meaning it's negative corelation. if 1, perfect corelation.
```

```python
import pandas as pd

df2 = df.copy()   #SNAS: to create a separate copy of df, because we don't want to change the value in df
#df2['tip_pct'] = df['tip'] / (df['total_bill'] * 100)   #SNAS: method 1. need to call directly the formula to execute to each rows under that column. otherwise, it will regard it as only 1 value to be put under a variable. no need to define a variable
#df2['tip_pct'] = pd.Series(df['tip'] / (df['total_bill'] * 100))   #SNAS: method 2. using pd.Series but cannot receive a single variable that save all the arguments. must run this pd.Series directly so that it can run throughout all data
df2['tip_pct'] = pd.Series(tip_pct)
df2.head()
```
