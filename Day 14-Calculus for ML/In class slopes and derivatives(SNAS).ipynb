```python
# e = X**4 - 80X**2 + 50x 

x = np.linspace(-10,10,10000)
y = (x)**4 - 80*(x**2) + 50*x

result = np.array(y)
plt.plot(x,result)

plt.axhline(0, color = 'black', linewidth = 0.4)
plt.axvline(0, color = 'black', linewidth = 0.4)
```
