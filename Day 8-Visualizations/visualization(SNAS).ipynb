```python
import matplotlib.pyplot as plt
```

```python
# create a bar plot
#SNAS: bar chart can takes make arguments

plt.bar(students, grades, color = 'pink')   #SNAS: to make bar chart # specify properties of our bar chart

plt.title('Chart of student grades')

plt.xlabel('student names')

plt.ylabel('marks')

#plt.show() #SNAS: if bar does not appear, try to run this code
```

```python
# create a scatter plot

plt.scatter(students, grades, color = 'pink')   #SNAS: to make scatter plot chart

plt.title('Chart of student grades')

plt.xlabel('student names')

plt.ylabel('marks')
```

```python
# all of these charts in the same place

# create a figure of 1 row and 1 columns
plt.figure (figsize = (15,5))   #SNAS: the pixel size for the display bar

#sub plot

plt.subplot(1,3,1)  # rows, columns, position
plt.bar(students, grades, color = 'pink')

#sub plot

plt.subplot(1,3,2)
plt.plot(students, grades, color = 'yellow')

#sub plot

plt.subplot(1,3,3)
plt.scatter(students, grades, color = 'blue')
```
