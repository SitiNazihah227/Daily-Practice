```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(30,10))
sns.scatterplot(data=ds, x='maker', y='state', hue='fuel')
plt.xticks(rotation=45)
```

```python
plt.figure(figsize=(30,10))
sns.countplot(data=ds, x="maker", order=ds["maker"].value_counts().index)
plt.title("Cars by Maker")
plt.xticks(rotation=45)
```

```python
# SNAS: Count cars grouped by fuel type and state
fuel_state = ds.groupby(['fuel', 'state']).size().unstack(fill_value=0)

# SNAS: Plot
fuel_state.plot(kind='bar', figsize = (15,6))
plt.title('Fuel Type vs State Comparison')
plt.xlabel('Fuel Type')
plt.ylabel('Number of Cars')
```
