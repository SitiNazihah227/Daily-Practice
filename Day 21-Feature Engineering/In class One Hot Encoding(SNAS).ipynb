```python
iris_onehot = iris.copy()

encoder = OneHotEncoder( sparse_output = False, handle_unknown='ignore')

catagorical_cols = ['species']

ecoded_values = encoder.fit_transform(iris_onehot[catagorical_cols]) # species

# names of new columns, that don't exist yet
new_columns = encoder.get_feature_names_out(catagorical_cols)
new_columns
# ['species_setosa', 'species_versicolor', 'species_virginica']

# iris_onehot

# adding the columns to the dataframe

iris_onehot[new_columns] = ecoded_values

iris_onehot.drop(columns=['species'], inplace = True)
iris_onehot
```

```python
titanic = sns.load_dataset('titanic')
titanic.head()
```

```python
titanic.info()  #SNAS: to know how many categorical dataset that need to be converted. Titanic categorical features: 9, numerical: 6
```

```python
titanic.sample(20)
```

```python
# SNAS: 1. label encoding

from sklearn.preprocessing import LabelEncoder

df = titanic.copy()

df['sex'] = df['sex'].astype('category')    #SNAS: to change datatype in pandas
#SNAS in above example, the species datatype will be category datatype

df['sex_cat_encoded'] = df['sex'].cat.codes #SNAS: species_cat_encoded is a new column
df
```

```python
#create LabelEncoder instance
encoder = LabelEncoder()

df['sex_sklearn_le'] = encoder.fit_transform(df['sex'])
df
```

```python
# SNAS: 2. one hot encoding

sex_onehot = titanic.copy()

encoder = OneHotEncoder(sparse_output = False, handle_unknown='ignore')

categorical_cols = ['sex']

ecoded_values = encoder.fit_transform(sex_onehot[categorical_cols]) # sex

# names of new columns, that don't exist yet
new_columns = encoder.get_feature_names_out(categorical_cols)
new_columns


# adding the columns to the dataframe
sex_onehot[new_columns] = ecoded_values

# remove the columns with object datatype sex
sex_onehot.drop(columns=['sex'], inplace = True)
sex_onehot
```
