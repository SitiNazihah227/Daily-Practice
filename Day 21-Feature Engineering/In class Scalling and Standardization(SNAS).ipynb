```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
```

```python
data = pd.DataFrame({
    'age' : [20, 50, 100],
    'income' : [20000, 50000, 100000]
})

#upscale/downscale
scaler = MinMaxScaler()

# Normalization
# you want to check what the data will look like in a scale of 0 to 1.
normalize_data = scaler.fit_transform(data)
normalize_data

```

```python
df_normalized = pd.DataFrame(normalize_data, columns=['age', 'income'])
df_normalized
```

- Why do we normalize data? why do we need to normalize data for machine algorithm
    - easier to calculate

- a lot of preference to salary over age in the given example
- our models only use numbers

```python
from sklearn.preprocessing import StandardScaler
```

```python
std_scaler = StandardScaler()

std_data = std_scaler.fit_transform(data)
std_data
# you got the z-score transformations for each data point
```

```python
df_std_data = pd.DataFrame(std_data, columns=['age', 'income'])
df_std_data
```
