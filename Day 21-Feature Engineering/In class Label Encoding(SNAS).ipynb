## FEATURE ENGINEERING ##

- It is the process of transforming raw data into features that make machine learning algorithms work better.
    - to improve performance of ML models
    - it can improve quality and relevance of features

**Categorical Features**
- Groups of data
    - will include Binary Features (0 and 1)
    - sex - M F
    - Survived - Yes No
    - Bool data type
    - Category data type - 
    - Ordinal data type - you can set up a scale in the cateogry - eg. small medium large, cold warm hot

**Numerical Features**
- where there are numbers
- will include 'float' and 'integers' both
- continous values
- grouping up will not make much sense (unless doing clustering)

**Encoding**

```python
# label encoding
# column that you can encode

penguins['species'].unique()

# {  }

species_codes = {
    'Adelie'    :   0,
    'Chinstrap' :   1,
    'Gentoo'    :   2
}

penguins['species_encoded'] = penguins['species'].map(species_codes)    #SNAS: passing a dictionary

penguins['species']
```

```python
# pandas map

s = pd.Series(['cat', 'cow', 'dog'])
print(s)

s2 = s.map({'cat' : 'kitten', 'cow' : 'calf', 'dog' : 'puppy'})
print(s2)
```
